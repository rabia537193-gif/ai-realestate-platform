```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

os.makedirs("assets", exist_ok=True)

# 1. Load Data
df = pd.read_csv("data/raw/train.csv")

# 2. EDA: Target Variable Distribution (Log Transformation)
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.histplot(df["SalePrice"], kde=True, color="blue")
plt.title("Original SalePrice Skewness")

plt.subplot(1, 2, 2)
sns.histplot(np.log1p(df["SalePrice"]), kde=True, color="green")
plt.title("Log-Transformed SalePrice (Normal)")
plt.tight_layout()
plt.savefig("assets/target_distribution.png")
plt.close()

# 3. Feature Relationship: Living Area vs Price & Outlier Removal
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x="GrLivArea", y="SalePrice", hue="OverallQual")
plt.title("GrLivArea vs SalePrice")
plt.savefig("assets/grlivarea_vs_price.png")
plt.close()

# 4. Correlation Matrix of Top Features
plt.figure(figsize=(8, 6))
top_cols = [
    "SalePrice",
    "OverallQual",
    "GrLivArea",
    "GarageCars",
    "TotalBsmtSF",
    "FullBath",
    "YearBuilt",
]
sns.heatmap(df[top_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Top Feature Correlations")
plt.tight_layout()
plt.savefig("assets/correlation_heatmap.png")
plt.close()

print("[EDA Complete] Visualizations saved in 'assets/' directory.")
```
